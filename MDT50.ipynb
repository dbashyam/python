````python
#Binary tree algorithm => searching

# 1. sorting (ascending order)
# 2. finding an middle value of list
# 3. compare middle value
# 4. if middle value == to_find result the output
# 5. if middle value > to_find go to left
# 6. if middle value < to_find go to right
````

````python
def binary_search(arr,target):
  low = 0
  high = len(arr) - 1
  while low <= high:
    mid = (low+high) //2
    if arr[mid] == target:
      return mid
    elif arr[mid] < target:
      low = mid + 1
    else:
      high = mid - 1
  return "not found"
````

````python
#Binary Tree search

class binary_tree:
  def __init__(self,data):
    self.data = data  #root node value
    self.left = None
    self.right = None
  def insert(self,new_data):
    if new_data < self.data:
      if self.left is None:
        self.left = binary_tree(new_data)
      else:
        self.left.insert(new_data)
    elif new_data > self.data:
      if self.right is None:
        self.right = binary_tree(new_data)
      else:
        self.right.insert(new_data)
    else:
      print("value already exist")

  def search(self,value):
    if value < self.data:
      if self.left is None:
        print("value not found")
      else:
        self.left.search(value)
    elif value > self.data:
      if self.right is None:
        print("value not found")
      else:
        self.right.search(value)
    else:
      print("value found")

#==========================================
#Only to show the tree data
#==========================================

  def show(self):
    if self.left:
      self.left.show()
    print(self.data)
    if self.right:
      self.right.show()
````

````python
#components to do hypothesis testing

- population and sample
- Estimation Technique
            - Point Estimation
            - Intervel Estimation
- Level of significance
- P value
````

````python
import numpy as np
import pandas as pd

population_size = 10000
population = np.random.normal(loc = 100,scale=15,size=population_size) #(loc=mean= 100 , scale=stdev=15 , size=no_of_datapoints=10000)
````

````python
population
````

````python
df_population = pd.DataFrame(population,columns=['population'])
````

````python
df_population.to_csv("population.csv")
````

````python
df_population
````

````python
import matplotlib.pyplot as plt

# Plot the population distribution
plt.figure(figsize=(10, 6))
plt.hist(population, bins=50, alpha=0.6, color='blue', label="Population")
plt.axvline(np.mean(population), color='red', linestyle='--', linewidth=2, label="Population Mean")
plt.title("Population Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.legend()
plt.show()
````

````python
df_population['population'].mean()
````

````python
sample_size = 30

sample = np.random.choice(population,size=sample_size,replace=False)

df_sample = pd.DataFrame(sample,columns=['sample_pop'])

df_sample
````

````python
df_sample['sample_pop'].mean()
````

````python
import scipy.stats as stats

population_mean = np.mean(population)

t_stat , p_value = stats.ttest_1samp(sample,population_mean)
````

````python
t_stat , p_value
````

````python
alpha = 0.05 #significance level

if p_value < alpha:
  print("Reject the Null Hypothesis (H0)")
else:
  print("Fail to reject the null hypothesis")
````

````python
# Plotting the t-distribution
x = np.linspace(-4, 4, 1000)
y = stats.t.pdf(x, df=sample_size - 1)  # t-distribution

# Critical t-values for two-tailed test
t_critical = stats.t.ppf(1 - alpha/2, df=sample_size - 1)

plt.figure(figsize=(10, 6))
plt.plot(x, y, label="T-Distribution", color="blue")

# Shade rejection regions
plt.fill_between(x, 0, y, where=(x <= -t_critical) | (x >= t_critical), color="red", alpha=0.4, label="Rejection Region")
plt.axvline(t_stat, color='green', linestyle='--', linewidth=2, label="Test Statistic")
plt.axvline(-t_critical, color='black', linestyle='--', label=f"Critical Value (-{t_critical:.2f})")
plt.axvline(t_critical, color='black', linestyle='--', label=f"Critical Value (+{t_critical:.2f})")
plt.title("T-Distribution with Rejection Regions")
plt.xlabel("T-Statistic")
plt.ylabel("Density")
plt.legend()
plt.show()
````

````python
import requests
import time
from IPython.display import display,clear_output

def get_iss_data():
  url = "http://api.open-notify.org/iss-now.json"

  response = requests.get(url)

  data = response.json()

  lat = data['iss_position']['latitude']
  lon = data['iss_position']['longitude']

  return lat,lon

for i in range(11):
  lat,lon = get_iss_data()

  m = folium.Map([lat,lon], zoom_start=12)

  folium.Marker(
      location=[lat,lon],
      tooltip="ISS Live data..!",
      popup=f"The current position {lat} {lon}",
      icon=folium.Icon(icon="cloud"),
  ).add_to(m)

  clear_output(wait=True)
  display(m)

  time.sleep(5)
````
